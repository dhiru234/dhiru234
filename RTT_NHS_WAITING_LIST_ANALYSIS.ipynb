```python



# Combine all three months
data_combined = pd.concat(
    [data_april, data_may, data_june],
    ignore_index=True
)

# Check the number of rows
print("April rows:", len(data_april))
print("May rows:", len(data_may))
print("June rows:", len(data_june))
print("Combined rows:", len(data_combined))

# Check the combined data
data_combined.head()
```

```python
# Remove leading/trailing spaces from column names
data_combined.columns = (
    data_combined.columns
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

print("Column names cleaned successfully.")
print(data_combined.columns.tolist())
```

```python
# Check missing values in every column

missing_values = data_combined.isnull().sum()

missing_summary = (
    missing_values[missing_values > 0]
    .sort_values(ascending=False)
)

print("Missing values by column:")
display(missing_summary)

print("\nTotal missing values:",
      data_combined.isnull().sum().sum())
```

```python
import numpy as np
```

```python
# ============================================================
# STEPS 5–10: CLEANING AND PREPROCESSING
# ============================================================

# 5. Clean text columns
text_columns = data_combined.select_dtypes(
    include=["object"]
).columns

for col in text_columns:
    data_combined[col] = data_combined[col].str.strip()

print("5. Text columns cleaned.")


# 6. Identify and convert date columns
date_columns = [
    col for col in data_combined.columns
    if "date" in col.lower()
]

print("\n6. Date columns found:")
print(date_columns)

for col in date_columns:
    data_combined[col] = pd.to_datetime(
        data_combined[col],
        errors="coerce"
    )

print("Date columns converted to datetime.")


# 7. Identify numeric columns
numeric_columns = data_combined.select_dtypes(
    include=["number"]
).columns.tolist()

print("\n7. Numeric columns:")
print(numeric_columns)


# 8. Identify and convert waiting-time columns
wait_columns = [
    col for col in data_combined.columns
    if "wait" in col.lower()
]

print("\n8. Waiting-time columns:")
print(wait_columns)

for col in wait_columns:
    data_combined[col] = pd.to_numeric(
        data_combined[col],
        errors="coerce"
    )

print("Waiting-time columns converted to numeric.")


# 9. Check for negative waiting times
print("\n9. Negative waiting-time check:")

for col in wait_columns:
    negative_values = (data_combined[col] < 0).sum()

    print(
        f"{col}: {negative_values} negative values"
    )


# 10. Descriptive statistics
print("\n10. Descriptive statistics:")

if len(wait_columns) > 0:
    display(
        data_combined[wait_columns].describe().T
    )
else:
    print("No waiting-time columns were detected.")

print("\nSteps 5–10 completed successfully.")
```

```python
# ============================================================
# STEPS 11–14: OUTLIERS, WAIT-TIME VALIDATION,
# FINAL QUALITY CHECK AND SAVE
# ============================================================

# ------------------------------------------------------------
# 11. IDENTIFY OUTLIERS USING IQR
# ------------------------------------------------------------

outlier_results = []

for col in wait_columns:

    Q1 = data_combined[col].quantile(0.25)
    Q3 = data_combined[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = (
        (data_combined[col] < lower_limit) |
        (data_combined[col] > upper_limit)
    ).sum()

    outlier_results.append({
        "Variable": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Number of Outliers": outliers
    })

outlier_summary = pd.DataFrame(outlier_results)

print("11. OUTLIER SUMMARY")
display(outlier_summary)

print(
    "\nOutliers have been identified but NOT removed."
)


# ------------------------------------------------------------
# 12. VALIDATE TOTAL WAIT TIME
# ------------------------------------------------------------

required_columns = [
    "Total Wait Time",
    "Registration Wait Time",
    "Triage Wait Time",
    "Professional Contact Wait Time"
]

print("\n12. WAIT-TIME VALIDATION")

# Check whether all required columns exist
missing_required = [
    col for col in required_columns
    if col not in data_combined.columns
]

if len(missing_required) == 0:

    # Calculate total from individual waiting phases
    data_combined["Calculated Total Wait Time"] = (
        data_combined["Registration Wait Time"] +
        data_combined["Triage Wait Time"] +
        data_combined["Professional Contact Wait Time"]
    )

    # Calculate difference
    data_combined["Wait Time Difference"] = (
        data_combined["Total Wait Time"] -
        data_combined["Calculated Total Wait Time"]
    )

    # Count mismatches
    mismatch_count = (
        data_combined["Wait Time Difference"] != 0
    ).sum()

    print(
        "Number of records with wait-time mismatch:",
        mismatch_count
    )

    print("\nWait-time difference summary:")
    display(
        data_combined["Wait Time Difference"].describe()
    )

else:

    print(
        "The following required columns were not found:"
    )

    print(missing_required)

    print(
        "\nWait-time validation was skipped."
    )


# ------------------------------------------------------------
# 13. FINAL DATA QUALITY CHECK
# ------------------------------------------------------------

print("\n==========================================")
print("13. FINAL DATA QUALITY CHECK")
print("==========================================")

print(
    "Total rows:",
    data_combined.shape[0]
)

print(
    "Total columns:",
    data_combined.shape[1]
)

print(
    "Total missing values:",
    data_combined.isnull().sum().sum()
)

print(
    "Total duplicate rows:",
    data_combined.duplicated().sum()
)

print(
    "Total numeric columns:",
    len(
        data_combined.select_dtypes(
            include=["number"]
        ).columns
    )
)

print(
    "Total date columns:",
    len(date_columns)
)

print(
    "Total waiting-time columns:",
    len(wait_columns)
)


# ------------------------------------------------------------
# 14. SAVE CLEANED AND PREPROCESSED DATA
# ------------------------------------------------------------

output_path = (
    "/content/drive/MyDrive/"
    "RTT-April-May-June-2026-Cleaned-Preprocessed.csv"
)

data_combined.to_csv(
    output_path,
    index=False
)

print("\n==========================================")
print("14. FILE SAVED SUCCESSFULLY")
print("==========================================")

print(output_path)
```

```python
# ============================================================
# REMAINING DATA PREPROCESSING
# ============================================================

# 1. Check missing values by column
missing_summary = (
    data_combined.isnull()
    .sum()
    .sort_values(ascending=False)
)

print("MISSING VALUES BY COLUMN")
display(missing_summary.head(30))


# ============================================================
# 2. Calculate missing percentage
# ============================================================

missing_report = pd.DataFrame({
    "Missing Values": data_combined.isnull().sum(),
    "Missing Percentage": (
        data_combined.isnull().sum() /
        len(data_combined) * 100
    )
})

missing_report = missing_report.sort_values(
    "Missing Percentage",
    ascending=False
)

print("\nMISSING DATA REPORT")
display(missing_report.head(30))


# ============================================================
# 3. Check columns with 100% missing values
# ============================================================

completely_empty = [
    col for col in data_combined.columns
    if data_combined[col].isna().all()
]

print("\nColumns with 100% missing values:")
print(completely_empty)

print(
    "Number of completely empty columns:",
    len(completely_empty)
)


# ============================================================
# 4. Remove only completely empty columns
# ============================================================

if len(completely_empty) > 0:
    data_combined = data_combined.drop(
        columns=completely_empty
    )

print(
    "\nColumns after removing completely empty columns:",
    data_combined.shape[1]
)


# ============================================================
# 5. Find columns with very high missingness
# ============================================================

high_missing = missing_report[
    missing_report["Missing Percentage"] > 50
]

print("\nColumns with more than 50% missing values:")
display(high_missing)


# ============================================================
# 6. Check possible RTT/waiting-time columns
# ============================================================

keywords = [
    "wait",
    "clock",
    "period",
    "days",
    "week",
    "breach",
    "pathway",
    "referral",
    "treatment",
    "admission",
    "appointment"
]

possible_rtt_columns = []

for col in data_combined.columns:
    if any(
        keyword in col.lower()
        for keyword in keywords
    ):
        possible_rtt_columns.append(col)

print("\nPossible RTT-related columns:")
print(possible_rtt_columns)


# ============================================================
# 7. Check categorical variables
# ============================================================

categorical_columns = data_combined.select_dtypes(
    include=["object"]
).columns.tolist()

print(
    "\nNumber of categorical columns:",
    len(categorical_columns)
)

print("\nCategorical columns:")
print(categorical_columns)


# ============================================================
# 8. Check unusual/invalid values in categorical variables
# ============================================================

for col in categorical_columns:

    unique_count = data_combined[col].nunique(
        dropna=True
    )

    print(
        f"{col}: {unique_count} unique values"
    )


# ============================================================
# 9. Check numeric variables for negative values
# ============================================================

numeric_columns = data_combined.select_dtypes(
    include=["number"]
).columns.tolist()

negative_summary = {}

for col in numeric_columns:

    negative_count = (
        data_combined[col] < 0
    ).sum()

    if negative_count > 0:
        negative_summary[col] = negative_count

print("\nNumeric columns containing negative values:")
print(negative_summary)


# ============================================================
# 10. Check infinite values
# ============================================================
import numpy as np
infinite_values = np.isinf(
    data_combined.select_dtypes(
        include=["number"]
    )
).sum().sum()

print(
    "\nTotal infinite numeric values:",
    infinite_values
)


# ============================================================
# 11. Replace infinite values with missing values
# ============================================================

data_combined.replace(
    [np.inf, -np.inf],
    np.nan,
    inplace=True
)

print(
    "Infinite values converted to NaN."
)


# ============================================================
# 12. Final duplicate check
# ============================================================

print(
    "\nFinal duplicate records:",
    data_combined.duplicated().sum()
)


# ============================================================
# 13. Final dataset dimensions
# ============================================================

print("\n======================================")
print("FINAL PREPROCESSING SUMMARY")
print("======================================")

print("Rows:", data_combined.shape[0])
print("Columns:", data_combined.shape[1])

print(
    "Missing cells:",
    data_combined.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    data_combined.duplicated().sum()
)


# ============================================================
# 14. SAVE PREPROCESSED DATA
# ============================================================

output_path = (
    "/content/drive/MyDrive/"
    "RTT-April-May-June-2026-Preprocessed.csv"
)

data_combined.to_csv(
    output_path,
    index=False
)

print("\nPreprocessed dataset saved:")
print(output_path)
```

# analyzing the backlog

```python
# ============================================================
# BACKLOG ANALYSIS
# ============================================================

# Identify RTT waiting-time columns
week_columns = [
    col for col in preprocess_path.columns
    if col.startswith("Gt ") and "Weeks" in col
]

# Convert waiting-time columns to numeric
for col in week_columns:
    preprocess_path[col] = pd.to_numeric(
        preprocess_path[col],
        errors="coerce"
    )

# ------------------------------------------------------------
# >18 WEEK BACKLOG
# ------------------------------------------------------------

backlog_18_columns = []

for col in week_columns:
    if col == "Gt 18 To 19 Weeks SUM 1":
        backlog_18_columns.append(col)

    elif col.startswith("Gt ") and "Weeks" in col:
        # Extract starting week number
        try:
            start_week = int(col.split("Gt ")[1].split(" ")[0])

            if start_week >= 19:
                backlog_18_columns.append(col)

        except:
            pass

# Add >18 week backlog
preprocess_path["Backlog_Over_18_Weeks"] = (
    preprocess_path[backlog_18_columns]
    .sum(axis=1, skipna=True)
)

# ------------------------------------------------------------
# >52 WEEK BACKLOG
# ------------------------------------------------------------

backlog_52_columns = []

for col in week_columns:
    try:
        start_week = int(col.split("Gt ")[1].split(" ")[0])

        if start_week >= 53:
            backlog_52_columns.append(col)

    except:
        pass

# Add >52 week backlog
preprocess_path["Backlog_Over_52_Weeks"] = (
    preprocess_path[backlog_52_columns]
    .sum(axis=1, skipna=True)
)

print("Backlog variables created successfully.")

print("Number of columns used for >18 weeks:",
      len(backlog_18_columns))

print("Number of columns used for >52 weeks:",
      len(backlog_52_columns))
# ============================================================
# OVERALL BACKLOG
# ============================================================

total_18_backlog = preprocess_path["Backlog_Over_18_Weeks"].sum()
total_52_backlog = preprocess_path["Backlog_Over_52_Weeks"].sum()

print("==========================================")
print("OVERALL RTT BACKLOG")
print("==========================================")

print(
    f"Total backlog >18 weeks: {total_18_backlog:,.0f}"
)

print(
    f"Total backlog >52 weeks: {total_52_backlog:,.0f}"
)
```

# Backlog by month

```python
# ============================================================
# MONTHLY BACKLOG
# ============================================================

monthly_backlog = (
    preprocess_path
    .groupby("Period")[
        ["Backlog_Over_18_Weeks", "Backlog_Over_52_Weeks"]
    ]
    .sum()
    .reset_index()
)

# Correct chronological order
month_order = [
    "RTT-April-2026",
    "RTT-May-2026",
    "RTT-June-2026"
]

monthly_backlog["Period"] = pd.Categorical(
    monthly_backlog["Period"],
    categories=month_order,
    ordered=True
)

monthly_backlog = (
    monthly_backlog
    .sort_values("Period")
    .reset_index(drop=True)
)

# Percentage change
monthly_backlog["18 Week Backlog % Change"] = (
    monthly_backlog["Backlog_Over_18_Weeks"]
    .pct_change() * 100
)

monthly_backlog["52 Week Backlog % Change"] = (
    monthly_backlog["Backlog_Over_52_Weeks"]
    .pct_change() * 100
)

display(monthly_backlog)
```

```python
# ============================================================
# GROUPED WAITING-TIME ANALYSIS
# ============================================================

import re

grouped_waiting = pd.DataFrame({
    "Period": preprocess_path["Period"].unique()
})

categories = [
    "0-18 Weeks",
    "19-26 Weeks",
    "27-52 Weeks",
    "53-78 Weeks",
    "79-104 Weeks",
    "Over 104 Weeks"
]

for category in categories:
    grouped_waiting[category] = 0

for col in week_columns:

    numbers = re.findall(r'\d+', col)

    if len(numbers) >= 2:
        start = int(numbers[0])
        end = int(numbers[1])

    elif len(numbers) == 1:
        start = int(numbers[0])
        end = 999

    else:
        continue

    if end <= 18:
        category = "0-18 Weeks"

    elif start >= 19 and end <= 26:
        category = "19-26 Weeks"

    elif start >= 27 and end <= 52:
        category = "27-52 Weeks"

    elif start >= 53 and end <= 78:
        category = "53-78 Weeks"

    elif start >= 79 and end <= 104:
        category = "79-104 Weeks"

    elif start >= 105:
        category = "Over 104 Weeks"

    else:
        continue

    values = (
        preprocess_path
        .groupby("Period")[col]
        .sum()
    )

    for period in values.index:
        grouped_waiting.loc[
            grouped_waiting["Period"] == period,
            category
        ] += values.loc[period]

display(grouped_waiting.round(0))
```

```python
# ============================================================
# PROVIDER OUTLIER ANALYSIS
# ============================================================

def find_outliers(df, column):

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    return outliers, lower_limit, upper_limit
```

```python
# ============================================================
# NATIONAL RTT KPI DASHBOARD
# April - June 2026
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Correct monthly order
# ------------------------------------------------------------

month_order = [
    "RTT-April-2026",
    "RTT-May-2026",
    "RTT-June-2026"
]

# ------------------------------------------------------------
# 2. NATIONAL DATA
# ------------------------------------------------------------
# Use the aggregate "Total" treatment function.
# Do NOT add Total together with individual specialties.

national_data = preprocess_path[
    (preprocess_path["Treatment Function Name"] == "Total") &
    (preprocess_path["Commissioner Org Code"] != "NONC")
].copy()

# ------------------------------------------------------------
# 3. Calculate monthly national KPIs
# ------------------------------------------------------------

national_kpi = (
    national_data
    .groupby("Period")
    .agg(
        Total_RTT_Volume=("RTT_Volume", "sum"),
        Backlog_Over_18_Weeks=("Backlog_Over_18_Weeks", "sum"),
        Backlog_Over_52_Weeks=("Backlog_Over_52_Weeks", "sum")
    )
    .reindex(month_order)
    .reset_index()
)

# ------------------------------------------------------------
# 4. Calculate pathways within 18 weeks
# ------------------------------------------------------------

national_kpi["Within_18_Weeks"] = (
    national_kpi["Total_RTT_Volume"]
    - national_kpi["Backlog_Over_18_Weeks"]
)

# ------------------------------------------------------------
# 5. Calculate percentages
# ------------------------------------------------------------

national_kpi["Within_18_Weeks_%"] = (
    national_kpi["Within_18_Weeks"]
    / national_kpi["Total_RTT_Volume"]
    * 100
)

national_kpi["Over_18_Weeks_%"] = (
    national_kpi["Backlog_Over_18_Weeks"]
    / national_kpi["Total_RTT_Volume"]
    * 100
)

national_kpi["Over_52_Weeks_%"] = (
    national_kpi["Backlog_Over_52_Weeks"]
    / national_kpi["Total_RTT_Volume"]
    * 100
)

# ------------------------------------------------------------
# 6. Monthly percentage changes
# ------------------------------------------------------------

national_kpi["RTT Volume % Change"] = (
    national_kpi["Total_RTT_Volume"]
    .pct_change() * 100
)

national_kpi[">18 Week Backlog % Change"] = (
    national_kpi["Backlog_Over_18_Weeks"]
    .pct_change() * 100
)

national_kpi[">52 Week Backlog % Change"] = (
    national_kpi["Backlog_Over_52_Weeks"]
    .pct_change() * 100
)

# April = baseline, therefore no percentage change
national_kpi.loc[
    national_kpi["Period"] == "RTT-April-2026",
    [
        "RTT Volume % Change",
        ">18 Week Backlog % Change",
        ">52 Week Backlog % Change"
    ]
] = np.nan

# ------------------------------------------------------------
# 7. Round values
# ------------------------------------------------------------

national_kpi = national_kpi.round(2)

# ------------------------------------------------------------
# 8. Display national KPI table
# ------------------------------------------------------------

display(national_kpi)
```

```python
# ============================================================
# RTT PROVIDER EFFICIENCY ANALYSIS
# ============================================================

import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. USE THE PROVIDER DATA ALREADY CREATED
# ------------------------------------------------------------

efficiency = provider.copy()

# Add backlog information
backlog_provider = (
    preprocess_path
    .groupby("Provider Org Name")[
        ["Backlog_Over_18_Weeks", "Backlog_Over_52_Weeks"]
    ]
    .sum()
    .reset_index()
)

efficiency = efficiency.reset_index()

efficiency = efficiency.merge(
    backlog_provider,
    on="Provider Org Name",
    how="left"
)

# ------------------------------------------------------------
# 2. CALCULATE EFFICIENCY INDICATORS
# ------------------------------------------------------------

# Incomplete pathways for every completed pathway
efficiency["Incomplete per Completed"] = (
    efficiency["Incomplete"] /
    efficiency["Completed"].replace(0, np.nan)
)

# >18-week backlog as % of total RTT volume
efficiency[">18 Week Burden (%)"] = (
    efficiency["Backlog_Over_18_Weeks"] /
    efficiency["Total RTT"] * 100
)

# >52-week backlog as % of total RTT volume
efficiency[">52 Week Burden (%)"] = (
    efficiency["Backlog_Over_52_Weeks"] /
    efficiency["Total RTT"] * 100
)

# ------------------------------------------------------------
# 3. CREATE EFFICIENCY SCORE
# ------------------------------------------------------------

# Higher completion rate = better
completion_score = (
    efficiency["Completion Rate (%)"]
    .rank(pct=True) * 100
)

# Lower backlog burden = better
backlog_score = (
    100 -
    efficiency[">18 Week Burden (%)"]
    .rank(pct=True) * 100
)

# Lower long-wait burden = better
long_wait_score = (
    100 -
    efficiency[">52 Week Burden (%)"]
    .rank(pct=True) * 100
)

# Combine the three indicators equally
efficiency["Efficiency Score"] = (
    completion_score * 0.50 +
    backlog_score * 0.30 +
    long_wait_score * 0.20
)

# ------------------------------------------------------------
# 4. EFFICIENCY RANK
# ------------------------------------------------------------

efficiency["Efficiency Rank"] = (
    efficiency["Efficiency Score"]
    .rank(ascending=False, method="min")
    .astype(int)
)

efficiency = efficiency.sort_values(
    "Efficiency Score",
    ascending=False
)

# ------------------------------------------------------------
# 5. DISPLAY RESULTS
# ------------------------------------------------------------

efficiency_table = efficiency[
    [
        "Provider Org Name",
        "Completed",
        "Incomplete",
        "Total RTT",
        "Completion Rate (%)",
        "Backlog_Over_18_Weeks",
        "Backlog_Over_52_Weeks",
        ">18 Week Burden (%)",
        ">52 Week Burden (%)",
        "Incomplete per Completed",
        "Efficiency Score",
        "Efficiency Rank"
    ]
].copy()

display(
    efficiency_table.head(20).round(2)
)
```
